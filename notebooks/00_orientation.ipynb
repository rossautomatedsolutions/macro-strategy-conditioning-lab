Question: What is this repository for and what is it not for?

**Warning:** This notebook is descriptive only and must not be used for forecasting, signal generation, or trading decisions.


```python
from pathlib import Path
import sys

import pandas as pd


def _repo_root() -> Path:
    for candidate in [Path.cwd(), *Path.cwd().parents]:
        if (candidate / "src").exists() and (candidate / "data").exists():
            return candidate
    raise RuntimeError("Could not locate repository root containing 'src' and 'data'.")


REPO_ROOT = _repo_root()
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

from src.data.loaders import load_regimes, load_strategy_returns, load_transitions
from src.data.validators import run_all_validations

real_regimes = load_regimes(REPO_ROOT / "data" / "macro" / "regimes.csv")
real_transitions = load_transitions(REPO_ROOT / "data" / "macro" / "transitions.csv")
example_regimes = load_regimes(REPO_ROOT / "data" / "macro" / "regimes_example.csv")
example_transitions = load_transitions(REPO_ROOT / "data" / "macro" / "transitions_example.csv")
narrow_strategy = load_strategy_returns(REPO_ROOT / "data" / "strategies" / "strategy_returns_example.csv")
wide_strategy = load_strategy_returns(REPO_ROOT / "data" / "strategies" / "strategy_returns_wide_sample.csv")

run_all_validations(example_regimes, example_transitions, narrow_strategy)
run_all_validations(real_regimes, real_transitions, narrow_strategy)
run_all_validations(real_regimes, real_transitions, wide_strategy)

scope = pd.DataFrame(
    [
        {"topic": "purpose", "detail": "Describe how strategy return streams behave across realized macro regimes and around realized regime transitions."},
        {"topic": "canonical_macro_inputs", "detail": "Use data/macro/regimes.csv and data/macro/transitions.csv from Part I for the 05-08 research path."},
        {"topic": "sample_strategy_inputs", "detail": "Use the narrow checked-in strategy fixture by default; swap to the wide fixture only for broader synthetic coverage."},
        {"topic": "out_of_scope", "detail": "Forecasting, signal generation, optimization, allocation, ranking, dashboards, and trading decisions remain out of scope."},
    ]
)
display(scope)

workflow = pd.DataFrame(
    [
        {"step": "00", "notebook": "orientation", "default_inputs": "real macro plus sample fixtures inventory", "purpose": "Confirm scope, constraints, and checked-in assets."},
        {"step": "01", "notebook": "inputs_validation", "default_inputs": "example macro + narrow strategy", "purpose": "Validate schemas and date coverage before any analysis."},
        {"step": "02-04", "notebook": "example descriptive MVP", "default_inputs": "example macro + narrow strategy", "purpose": "Show regime, risk, and transition diagnostics on the example contract fixtures."},
        {"step": "05-08", "notebook": "canonical Part II flow", "default_inputs": "real macro + narrow strategy", "purpose": "Run the real Part I handoff path through the final descriptive summary packet."},
    ]
)
display(workflow)

artifact_summary = pd.DataFrame(
    [
        {
            "artifact": "real_macro_regimes",
            "path": "data/macro/regimes.csv",
            "rows": len(real_regimes),
            "start": real_regimes["DATE"].min(),
            "end": real_regimes["DATE"].max(),
            "notes": f"{real_regimes['Macro_Regime'].nunique()} realized macro regime categories",
        },
        {
            "artifact": "real_macro_transitions",
            "path": "data/macro/transitions.csv",
            "rows": len(real_transitions),
            "start": real_transitions["transition_date"].min(),
            "end": real_transitions["transition_date"].max(),
            "notes": "canonical transition event table for 05-08",
        },
        {
            "artifact": "example_strategy_fixture",
            "path": "data/strategies/strategy_returns_example.csv",
            "rows": len(narrow_strategy),
            "start": narrow_strategy["DATE"].min(),
            "end": narrow_strategy["DATE"].max(),
            "notes": "default narrow sample fixture; zero aligned transitions is acceptable",
        },
        {
            "artifact": "wide_strategy_fixture",
            "path": "data/strategies/strategy_returns_wide_sample.csv",
            "rows": len(wide_strategy),
            "start": wide_strategy["DATE"].min(),
            "end": wide_strategy["DATE"].max(),
            "notes": "optional broader synthetic coverage for diagnostics and tests",
        },
    ]
)
display(artifact_summary)

validation_summary = pd.DataFrame(
    [
        {"validated_path": "example macro + narrow strategy", "status": "PASS"},
        {"validated_path": "real macro + narrow strategy", "status": "PASS"},
        {"validated_path": "real macro + wide strategy", "status": "PASS"},
    ]
)
display(validation_summary)
```
